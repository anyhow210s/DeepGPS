# Data Exploration

**English:** Explore the local Google Takeout photo folders before training. Count images by location and display representative samples. JSON files and videos are intentionally excluded.

**한국어:** 학습 전에 로컬 Google Takeout 사진 폴더를 탐색합니다. 위치별 이미지 수를 확인하고 대표 샘플을 표시합니다. JSON 파일과 동영상은 의도적으로 제외합니다.

```python
from pathlib import Path
from collections import Counter
import matplotlib.pyplot as plt
from PIL import Image

PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
RAW_DIR = PROJECT_ROOT / 'data' / 'raw'
EXTENSIONS = {'.jpg', '.jpeg', '.png', '.heic', '.heif'}
```

```python
class_images = {
    class_dir.name: [p for p in class_dir.iterdir() if p.is_file() and p.suffix.lower() in EXTENSIONS]
    for class_dir in sorted(RAW_DIR.iterdir()) if class_dir.is_dir()
}
counts = Counter({name: len(paths) for name, paths in class_images.items()})
counts
```

```python
plt.figure(figsize=(10, 4))
plt.bar(counts.keys(), counts.values())
plt.title('Raw Image Count by Location')
plt.ylabel('Images')
plt.xticks(rotation=30, ha='right')
plt.tight_layout();
```

```python
fig, axes = plt.subplots(1, max(len(class_images), 1), figsize=(4 * max(len(class_images), 1), 4))
axes = [axes] if len(class_images) <= 1 else axes
for axis, (class_name, paths) in zip(axes, class_images.items()):
    if paths:
        try:
            axis.imshow(Image.open(paths[0]).convert('RGB'))
        except Exception as error:
            axis.text(0.5, 0.5, str(error), ha='center')
    axis.set_title(class_name)
    axis.axis('off')
plt.tight_layout();
```
