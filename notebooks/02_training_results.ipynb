# Training Results

**English:** Review learning curves, classification metrics, and the confusion matrix after running `train.py` and `evaluate.py`.

**한국어:** `train.py`와 `evaluate.py` 실행 후 학습 곡선, 분류 성능 지표, 혼동 행렬을 확인합니다.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
from PIL import Image

PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
REPORTS_DIR = PROJECT_ROOT / 'outputs' / 'reports'
FIGURES_DIR = PROJECT_ROOT / 'outputs' / 'figures'
```

```python
history = pd.read_csv(REPORTS_DIR / 'training_history.csv')
history
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
axes[0].plot(history['epoch'], history['train_loss'], label='Training')
axes[0].plot(history['epoch'], history['val_loss'], label='Validation')
axes[0].set_title('Loss')
axes[0].legend()
axes[1].plot(history['epoch'], history['train_accuracy'], label='Training')
axes[1].plot(history['epoch'], history['val_accuracy'], label='Validation')
axes[1].set_title('Accuracy')
axes[1].legend()
plt.tight_layout();
```

```python
report = pd.read_csv(REPORTS_DIR / 'classification_report.csv', index_col=0)
report
```

```python
confusion_matrix_image = Image.open(FIGURES_DIR / 'confusion_matrix.png')
plt.figure(figsize=(9, 7))
plt.imshow(confusion_matrix_image)
plt.axis('off');
```
